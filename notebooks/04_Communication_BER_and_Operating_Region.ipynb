# Communication BER and operating-region resultsLoads the committed BER curves and shows the pooled operating-region BER (SNR >= 5 dB) together with the 1e-4 floor for the three swarm scenarios, reproducing Table II of the paper.

```python
import sys
from pathlib import Path
import numpy as np
sys.path.insert(0, str(Path.cwd()))
```

## Load the committed BER curves

```python
import pandas as pd
curves = pd.read_csv('results/curves/ber_curves.csv')
print(curves.head())
```

## Pooled BER over the operating region (SNR >= 5 dB)

```python
def pooled(group):
    m = group.snr_db >= 5
    return (group.ber[m] * group.n_symbols[m]).sum() / group.n_symbols[m].sum()

print(curves.groupby(['scenario', 'model']).apply(pooled).unstack())
```

## Minimum SNR at which the BER reaches 1e-4

```python
def floor_snr(group):
    hit = group[group.ber <= 1e-4]
    return hit.snr_db.min() if len(hit) else float('nan')

print(curves.groupby(['scenario', 'model']).apply(floor_snr).unstack())
```

The figure is `results/figures/operating_region_ber.pdf` (Figure 2 of the paper).
