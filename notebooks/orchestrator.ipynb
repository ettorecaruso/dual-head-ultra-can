```python
import os, subprocess
REPO = '/content/dual-head-ultra-can'
URL = 'https://github.com/ettorecaruso/dual-head-ultra-can.git'
if not os.path.isdir(os.path.join(REPO, '.git')):
    subprocess.run(['git', 'clone', '-q', URL, REPO], check=True)
else:
    subprocess.run(['git', '-C', REPO, 'pull', '-q', '--ff-only'], check=True)
os.chdir(REPO)
commit = subprocess.run(['git', '-C', REPO, 'rev-parse', '--short', 'HEAD'],
                        capture_output=True, text=True).stdout.strip()
print('Repo:', REPO, '| commit:', commit)
```
